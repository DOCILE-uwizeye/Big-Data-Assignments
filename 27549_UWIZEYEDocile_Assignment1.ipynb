## Task 1: Student Profile System (20 marks)

```python
# Task 1: Student Profile System
# Collect student information and display a formatted profile

print('===== STUDENT PROFILE SYSTEM =====')

# Collect all inputs from the user
full_name       = input('Enter your full name: ')
student_id      = input('Enter your student ID: ')
age             = int(input('Enter your age: '))
country         = input('Enter your country of origin: ')
year_of_study   = int(input('Enter your current year of study (1-4): '))
gpa             = float(input('Enter your current GPA (0.0 - 4.0): '))
major           = input('Enter your major/course of study: ')

# Calculate expected graduation year
current_year        = 2025
remaining_years     = 4 - year_of_study
graduation_year     = current_year + remaining_years

# Display formatted student profile using f-strings
print()
print('=' * 45)
print('           STUDENT PROFILE CARD')
print('=' * 45)
print(f'  Full Name       : {full_name}')
print(f'  Student ID      : {student_id}')
print(f'  Age             : {age} years')
print(f'  Country         : {country}')
print(f'  Year of Study   : Year {year_of_study}')
print(f'  GPA             : {gpa:.2f}')
print(f'  Major           : {major}')
print(f'  Graduation Year : {graduation_year}')
print('=' * 45)
```

## Task 4: String Processing Pipeline (5 marks)

```python
# Task 4: String Processing Pipeline
# Extract and process information from a raw data string

raw_data = '   MUGISHA Jean Pierre | 1998 | Kigali, Rwanda | Computer Science | 3.72   '

# Step 1: Strip leading and trailing whitespace
cleaned = raw_data.strip()
print(f'Stripped: {cleaned}')

# Step 2: Split by | separator
fields = cleaned.split('|')

# Step 3: Strip whitespace from each field
fields = [field.strip() for field in fields]
print(f'Fields: {fields}')

# Step 4: Extract and convert each field
full_name_raw  = fields[0]                          # raw name
birth_year     = int(fields[1])                     # birth year as integer
location       = fields[2]                          # city, country
major          = fields[3]                          # major
gpa            = float(fields[4])                   # GPA

# Convert name to title case
full_name = full_name_raw.title()

# Split full name into first and last name
name_parts = full_name.split()
last_name  = name_parts[0]                          # MUGISHA
first_name = ' '.join(name_parts[1:])               # Jean Pierre
full_name_combined = first_name + ' ' + last_name   # concatenated

# Calculate age
current_year = 2025
age = current_year - birth_year

# Extract city and country separately
location_parts = location.split(',')
city    = location_parts[0].strip()
country = location_parts[1].strip()

# Check if 'Science' appears in the major
has_science = 'Science' in major

# Print formatted summary
print()
print('=' * 40)
print('       EXTRACTED PROFILE SUMMARY')
print('=' * 40)
print(f'  Full Name    : {full_name}')
print(f'  First Name   : {first_name}')
print(f'  Last Name    : {last_name}')
print(f'  Full Name    : {full_name_combined}')
print(f'  Birth Year   : {birth_year}')
print(f'  Age          : {age} years')
print(f'  City         : {city}')
print(f'  Country      : {country}')
print(f'  Major        : {major}')
print(f'  GPA          : {gpa}')
print(f'  Has Science  : {has_science}')
print('=' * 40)
```

## Task 5: Financial Calculator (5 marks)

```python
# Task 5: Financial Calculator
# Calculate simple interest, compound interest, expenses and net savings

print('===== FINANCIAL CALCULATOR =====')

# Collect inputs from the user
principal         = int(float(input('Enter principal amount (RWF): ')))
annual_rate       = float(input('Enter annual interest rate (e.g. 12 for 12%): '))
years             = int(input('Enter number of years: '))
monthly_expense   = float(input('Enter monthly expense amount (RWF): '))

# --- Calculations ---

# Simple Interest: I = P x R x T / 100
simple_interest       = principal * annual_rate * years / 100
simple_total          = principal + simple_interest

# Compound Interest: A = P x (1 + R/100) ^ T
compound_total        = principal * (1 + annual_rate / 100) ** years
compound_interest     = compound_total - principal

# Difference between compound and simple interest
interest_difference   = compound_total - simple_total

# Total expenses over the full period
total_expenses        = monthly_expense * 12 * years

# Net savings after expenses
net_savings           = compound_total - total_expenses

# --- Display Results ---
print()
print('=' * 50)
print('           FINANCIAL SUMMARY REPORT')
print('=' * 50)
print(f'  Principal Amount       : {principal:>15,.2f} RWF')
print(f'  Annual Rate            : {annual_rate:>14.2f}%')
print(f'  Period                 : {years:>12} years')
print(f'  Monthly Expense        : {monthly_expense:>15,.2f} RWF')
print('-' * 50)
print(f'  Simple Interest        : {simple_interest:>15,.2f} RWF')
print(f'  Total (Simple)         : {simple_total:>15,.2f} RWF')
print('-' * 50)
print(f'  Compound Interest      : {compound_interest:>15,.2f} RWF')
print(f'  Total (Compound)       : {compound_total:>15,.2f} RWF')
print('-' * 50)
print(f'  Interest Difference    : {interest_difference:>15,.2f} RWF')
print(f'  Total Expenses         : {total_expenses:>15,.2f} RWF')
print(f'  Net Savings            : {net_savings:>15,.2f} RWF')
print('=' * 50)
```
